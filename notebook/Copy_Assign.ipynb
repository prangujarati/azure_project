```python
import subprocess
import logging
import os

# Set up logging to stdout (will show up in logs when run as a job)
logging.basicConfig(level=logging.INFO)
logger = logging.getLogger("wheel_installer")

# Get the .whl path from widget
dbutils.widgets.text(
    name="package_path",
    defaultValue="/dbfs/FileStore/packages/Assign/default_package.whl",
    label="Enter the path to the .whl file:"
)

package_path = dbutils.widgets.get("package_path")

# Remove "/dbfs" prefix when using with pip
pip_path = package_path.replace("/dbfs", "")

# Check if the file exists first
if not os.path.exists(package_path):
    logger.error(f"Wheel file not found at: {package_path}")
else:
    try:
        logger.info(f"Installing package from: {pip_path}")
        result = subprocess.run(
            ["pip", "install", "--disable-pip-version-check", "--force-reinstall", pip_path],
            check=True,
            stdout=subprocess.PIPE,
            stderr=subprocess.PIPE,
            text=True
        )
        logger.info("Installation output:\n" + result.stdout)
    except subprocess.CalledProcessError as e:
        logger.error("Package installation failed!")
        logger.error("STDOUT:\n" + e.stdout)
        logger.error("STDERR:\n" + e.stderr)
```
